```python
df['release_year'] =df['release_year'].astype('int64')
df['date_added'] = df['date_added'].astype('datetime64[ns]')
```

```python
df['duration_value'] = df['duration_value'].fillna(0)
```

```python
df['duration_value'] = df['duration_value'].astype('int64')
```

```python
df.isna().sum()
```
